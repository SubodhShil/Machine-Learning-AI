```python
import re 
text = "Hello, world! This is a test."
result = re.split(r'(\s)', text)

print(result)
```

```python
special_characters = [',', '.', '!', '?', ';', ':', '-', "--", "\\", "/", '_', '(', ')', '[', ']', '{', '}', '"', "'"]

# Escape each special character and join them into a character class
pattern = r'([' + re.escape(''.join(special_characters)) + r']|\s)'

text = "Hello, world! How's everything?"
result = [t for t in re.split(pattern, text) if t and not t.isspace()]
```
